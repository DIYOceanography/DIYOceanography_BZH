## Background

### Introduction 

Under specific illumination (usually UV), some molecules named fluorophore emit visible light. The color is different depending on the fluorophore (Figure [1](#intro_fluo)). 
Fluorescence is the ability of a molecule to a emit light of a different wavelength than the one used to lighten it.

### General principle 

Fluorescence is a nonlinear optical phenomenon where new wavelengths are produced from melocular excition (Figure [2](#intro_physic)). When a photon with a specific excitation wavelength $\lambda_{Exc}$ is absorbed by a molecule, it promotes the molecule to a higher energy state. From this excited state, non-radiative and radiative transitions spontaneously occur to reduce the molecule's energy level (Figure [2](#physic_Jablonski)). The non-radiatives transition introduce a time delay between excitation and emission referred as the excited lifetime $\tau$ and is typically measured to characterized a fluorophore. The radiative transition generates light emission at lower energy, corresponding to a longer wavelength $\lambda_{Emit}$. The energy shift, called the Stokes shift, is important to the fluorescent technique sensititivity. Note that in reality the molecule are sensitive to a spectrum of light and emit a spectrum of wavelength. For ease of writing we keep the notion of unique wavelength. Fluorescence refers to these three stage, illustrated in Jablonski diagrams.

```{figure} ../_static/fluo/fluo_intro.png

Figure 2 - Scheme representing linear (top) and nonlinear (bottom) optics. An "Eperlan du large" fish (middle picture) receives light at given wavelengths λ (left picture and colored arrow). The emitted light spectrum (right colored arrow) depends on the incident light. The last two parameters used in fluorescent technique are the quantum yielf and the molar absorption coefficient. In a fluorescent material, only a portion of the excited fluorophore produce fluorescence emission as other processes may also depopulate the excited state. The ratio between number of fluorescence photons emitted to the number of photons absorbed measure these processes and is the quantum yield. The molar absorption coefficient quantify the light absorbed by the fluorophore. Learn more about fluorescence on the [Molecular Probes Handbook](https://www.thermofisher.com/fr/fr/home/references/molecular-probes-the-handbook/introduction-to-fluorescence-techniques.html).
```

To summarize, the key parameters include:
- Excitation spectrum
- Emission spectrum
- Molar absorption coefficient 
- Quantum yield
- The Stokes shift
- Excited state lifetime  : $\tau$

These parameters depend on the physical and chemical characteristics of the fluorescent material and its environment (pH, Cl⁻ concentration, temperature, CO₂ levels, etc.). This environmental sensitivity makes fluorescence a valuable tool for studying living organisms and their surroundings.

### Proxy applications

Fluorescence serves important functions for marine life such as fish and sharks (https://www.nature.com/articles/srep24751), and has found widespread applications in human endeavors. Fluorescence measurements are used as proxies in medical diagnostics for cancer detection (https://bmccancer.biomedcentral.com/articles/10.1186/1471-2407-12-21), industrial processes, and research applications (fluorescence research has been recognized with two Nobel prizes: https://www.lemonde.fr/planete/article/2008/10/09/prix-nobel-la-fluorescence-d-une-meduse-primee_1105007_3244.html, https://presse.inserm.fr/cest-dans-lair/prix-nobel-de-chimie-2014/). 

In oceanography, biofluorescence is employed to study and differentiate phytoplankton populations and monitor coral health (https://www.nature.com/articles/srep01421). Cost-effective fluorescent substances (such as carbon dots) are used to track floating objects (https://www.nature.com/articles/s41598-025-17302-9) and serve as proxies for pH measurements (https://www.nature.com/articles/s41598-023-37578-z). Fluorescence from colored dissolved organic matter (CDOM) enables tracking of ocean currents and water mass identification: https://www.nature.com/articles/s41598-021-86948-y, https://www.nature.com/articles/srep33978. Different organic components can be distinguished by their unique fluorescence spectra, which proves useful for differentiating Pacific and riverine water masses (adapted from: https://www.nature.com/articles/s41598-021-86948-y).
