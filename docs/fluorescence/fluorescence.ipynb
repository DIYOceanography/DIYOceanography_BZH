Under specific illumination (usually UV), molecules called fluorophores emit visible light. The color varies depending on the fluorophore (Figure [1](#intro_fluo)). 
Fluorescence is the ability of a molecule to emit light at a different wavelength than the one used to excite it. 


### General principle 

Fluorescence is a nonlinear optical phenomenon where new wavelengths are produced from molecular excitation (Figure [2](#intro_physic)). When a photon with a specific excitation wavelength $\lambda_{Exc}$ is absorbed by a molecule, it promotes the molecule to a higher energy state. From this excited state, non-radiative and radiative transitions spontaneously occur to reduce the molecule's energy level (Figure [3](#physic_Jablonski)). The radiative transition generates light emission at lower energy, corresponding to a longer wavelength $\lambda_{Emit}$. Molecules capable of generating this radiative transition at different wavelengths are called fluorophores. Note that in reality, fluorophores are sensitive to a spectrum of light and emit a spectrum of wavelengths. 

```{figure} ../_static/fluo/fluo_intro.png
:width: 500px
:name: intro_physic
:align: left

Figure 2 - Scheme representing linear (top) and nonlinear (bottom) optics. An Eperlan du large (middle picture) receives light at given wavelengths λ (left picture and colored arrow). The emitted light spectrum (right colored arrow) depends on the incident light. In linear optics, the fish receives visible wavelengths (400-900 nm) and reflects light with the same spectral components and different intensity. Here, the fish has absorbed red, orange, and yellow wavelengths, making it appear purple, blue and green. In nonlinear optics, the fish receives UV light (λ < 400 nm) and emits both reduced UV and green light; this phenomenon is fluorescence.
```

The energy shift, called the Stokes shift, can be important for some fluorescent technique sensitivity [The Molecular Probes® Handbook—A Guide to Fluorescent Probes and Labeling Technologies]. Fluorescence encompasses these three stages, illustrated by Jablonski diagrams (Figure [3](#physic_Jablonski)). The non-radiative transitions introduce a time delay between excitation and emission referred to as the excited lifetime $\tau$, which is typically measured to characterize a fluorophore.

All biological and most organic materials fluoresce when excited with UV light [Misra et al. 2025]. The intensity of fluorescence is not constant and depends on the physiological (i.e., physical and chemical) state of the fluorescent material and its environment (pH, Cl⁻ concentration, temperature, CO₂ levels, etc.). In a fluorescent material, only a portion of the excited molecules produce fluorescence emission as other processes may also depopulate the excited state. Quenching refers to any process that decreases fluorescence intensity [Lakowicz 2006]. The ratio between the number of fluorescence photons emitted to the number of photons absorbed measures the fluorescence efficiency and is referred to as quantum yield. Learn more about fluorescence in the [Molecular Probes Handbook](https://www.thermofisher.com/fr/fr/home/references/molecular-probes-the-handbook/introduction-to-fluorescence-techniques.html). In oceanography quenching processes are typicall used to measured envrionmental parameters such as oxygen or pH. Learn more about oxygen and pH measurements in the specific sections.

To summarize, the key parameters in fluorescence include:
- The fluorophore 
- Its excitation spectrum
- Its emission spectrum
- The excited state lifetime: $\tau$
- The quantum yield
- Molar absorption coefficient: relates the absorbance to the molecular concentration using Beer-Lambert's law

These parameters depend on the material and its environment. Biological and organic materials have different fluorescence spectra [Fig 2. in Misra et al. 2025] which makes a powerful tool to distinguish species. The intensity also depends on the physiological state of the living organism or on the number of times the sample fluoresces. Fluorescence intensity also depends on environemental parameter such as salinity, oxygen and pH through quenching processes. For instance the more dissolved oygen the less intensity is emitted by a givan sample of fluorophore, this specificity is exploited in optodes.  

### Proxy applications

Fluorescence is a powerful tool that has found widespread applications in human endeavors. Fluorescence measurements are used as proxies in medical diagnostics for cancer detection (https://bmccancer.biomedcentral.com/articles/10.1186/1471-2407-12-21), industrial processes, and research applications (fluorescence research has been recognized with two Nobel prizes: https://www.lemonde.fr/planete/article/2008/10/09/prix-nobel-la-fluorescence-d-une-meduse-primee_1105007_3244.html, https://presse.inserm.fr/cest-dans-lair/prix-nobel-de-chimie-2014/). 

In oceanography, fluorescence serves important functions for marine life such as fish and sharks (https://www.nature.com/articles/srep24751), and biofluorescence is employed to study and differentiate phytoplankton populations and monitor coral health (https://www.nature.com/articles/srep01421). Fluorescent cameras are used to track floating and submerged objects (https://www.nature.com/articles/s41598-025-17302-9) and serve as proxies for pH measurements (https://www.nature.com/articles/s41598-023-37578-z). Fluorescence from colored dissolved organic matter (CDOM) enables tracking of ocean currents and water mass identification: https://www.nature.com/articles/s41598-021-86948-y, https://www.nature.com/articles/srep33978. Different organic components can be distinguished by their unique fluorescence spectra, which proves useful for differentiating Pacific and riverine water masses (adapted from: https://www.nature.com/articles/s41598-021-86948-y). 

### Units and standards

Calibration of fluorescence sensors in oceanographic applications requires rigorous adherence to established standards to ensure data comparability across instruments and research campaigns. For chlorophyll-a fluorescence measurements, primary calibration is typically performed using extracted chlorophyll solutions of known concentration, prepared from cultured phytoplankton or spinach extracts dissolved in acetone (Holm-Hansen et al., 1965; Lorenzen, 1966). Secondary standards include solid-state references or stable fluorescent dyes such as Rhodamine WT, which provides long-term stability for field deployments (Smart et al., 1996; Carlson et al., 2017). CDOM fluorescence is conventionally calibrated against quinine sulfate dihydrate solutions, with results expressed in Quinine Sulfate Units (QSU) or parts per billion (ppb) quinine sulfate equivalents. Instrument-specific calibration coefficients must account for temperature dependencies, non-linear response at high concentrations, and potential quenching effects. Regular verification using certified reference materials and inter-comparison exercises between instruments are essential practices to maintain measurement accuracy and facilitate integration of datasets across different platforms and temporal scales.
