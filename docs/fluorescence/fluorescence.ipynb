## Background
```{figure} ../_static/fluo/fluo_intro_0.png
:width: 400px
:name: intro_fluo
:align: center

Figure 1 - Different solutions with fluorophores
```
Under specific illumination (usually UV), molecules called fluorophores emit visible light. The color varies depending on the fluorophore (Figure [1](#intro_fluo)). 
Fluorescence is the ability of a molecule to emit light at a different wavelength than the one used to excite it. 


### General principle 

Fluorescence is a nonlinear optical phenomenon where new wavelengths are produced from molecular excitation (Figure [2](#intro_physic)). When a photon with a specific excitation wavelength $\lambda_{Exc}$ is absorbed by a molecule, it promotes the molecule to a higher energy state. From this excited state, non-radiative and radiative transitions spontaneously occur to reduce the molecule's energy level (Figure [3](#physic_Jablonski)). The radiative transition generates light emission at lower energy, corresponding to a longer wavelength $\lambda_{Emit}$ than $\lambda_{Exc}$. These molecules are called fluorophores. 

```{figure} ../_static/fluo/fluo_intro.png
:width: 500px
:name: intro_physic
:align: left

Figure 2 - Scheme representing linear (top) and nonlinear (bottom) optics. An Eperlan du large (middle picture) receives light at given wavelengths λ (left picture and colored arrow). The emitted light spectrum (right colored arrow) depends on the incident light. In linear optics, the fish receives visible wavelengths (400-900 nm) and reflects light with the same spectral components and different intensity. Here, the fish has absorbed red, orange, and yellow wavelengths, making it appear purple, blue and green. In nonlinear optics, the fish receives UV light (λ < 400 nm) and emits both reduced UV and green light; this phenomenon is fluorescence.
```

The energy shift, called the Stokes shift, can be important for fluorescence technique sensitivity [The Molecular Probes® Handbook—A Guide to Fluorescent Probes and Labeling Technologies]. Fluorescence encompasses these three stages, illustrated by Jablonski diagrams (Figure [3](#physic_Jablonski)). The non-radiative transitions introduce a time delay between excitation and emission referred to as the excited state lifetime $\tau$, which is typically measured to characterize a fluorophore. 


```{figure} ../_static/fluo/fluo_Jablonski.svg
:width: 250px
:name: physic_Jablonski
:align: right

Figure 3 - Jablonski diagram of absorbance, non-radiative decay, and fluorescence. Electronic transitions are about 1 eV. Absorption is about 1 femtosecond, relaxation takes about 1 picosecond, fluorescence takes about 1 nanosecond. S0 and S1 represent different electronic states. The other numbers (here 0–3 are shown) represent vibrational states. By Jacobkhed from the [Jablonski Wikipedia page](https://fr.wikipedia.org/wiki/Diagramme_de_Jablonski) (french version).
```

All biological and most organic materials fluoresce when excited with UV light [Misra et al. 2025]. The intensity of fluorescence is not constant and depends on the physiological (i.e., physical and chemical) state [Gorbunov et Falkowski 2022] of the fluorescent material and its environment (pH, Cl⁻ concentration, temperature, CO₂ levels, etc.). In a fluorescent material, only a portion of the excited molecules produce fluorescence emission as other processes may also depopulate the excited state. Quenching refers to any process that decreases fluorescence intensity [Lakowicz 2006]. The ratio between the number of fluorescence photons emitted to the number of photons absorbed measures the fluorescence efficiency and is referred to as quantum yield. Learn more about fluorescence in the [Molecular Probes Handbook](https://www.thermofisher.com/fr/fr/home/references/molecular-probes-the-handbook/introduction-to-fluorescence-techniques.html). In oceanography, quenching processes are typically used to measure environmental parameters such as oxygen or pH. Learn more about oxygen and pH measurements in the specific sections.


To summarize, the five key parameters measured or exploited are:
- The excitation spectrum and the maximal excitation wavelength
- The emission spectrum and the maximal emission wavelength 
- The excited state lifetime: $\tau$
- The quantum yield
- Molar absorption coefficient: relates the absorbance to the molecular concentration using Beer-Lambert's law


### Proxy applications

Fluorescence is exploited across multiple disciplines, from [cancer detection](https://bmccancer.biomedcentral.com/articles/10.1186/1471-2407-12-21) to quality control in industrial processes. Fluorescence has led to two Nobel prizes: one for [green fluorescent protein (GFP) applications](https://www.lemonde.fr/planete/article/2008/10/09/prix-nobel-la-fluorescence-d-une-meduse-primee_1105007_3244.html) and one in 2014 for [high-resolution fluorescence microscopy](https://presse.inserm.fr/cest-dans-lair/prix-nobel-de-chimie-2014/).

In oceanography, fluorescence is used as a [proxy (statistics)](https://en.wikipedia.org/wiki/Proxy_(statistics)#:~:text=In%20statistics%2C%20a%20proxy%20or,with%20the%20variable%20of%20interest.) for unobservable or immeasurable variables in biology, physics, and chemistry. 

#### Biological oceanography applications
- **Biofluorescence** serves important functions in marine ecosystems, with natural biofluorescence observed in fish, sharks, and other marine organisms. This natural fluorescence aids in species identification and behavioral studies [Gruber et al. 2016].

- **Phytoplankton biomass** is one of the [essential ocean variables (EOV)](https://goosocean.org/document/17507) studied with fluorimetry. Fluorescence is correlated with Chlorophyll-a, which is a proxy for phytoplankton biomass [Lorenzen 1966]. Different phytoplankton species exhibit distinct fluorescence signatures, enabling community structure analysis and species differentiation [Gorbunov et Falkowski 2022; Harris et al. 2024].

- **Coral fluorescence patterns** serve as indicators of coral stress and health. Changes in fluorescence intensity and spectral characteristics can detect early signs of environmental stress before visible symptoms appear [Roth et Deheyn 2013].

#### Tracer and water mass studies

- **Fluorescent dye** tracing techniques are widely used in hydrology to map the motion of rivers and ocean currents with fluorometers [Wilson et al.]; [Park et al. 2023]. Their usefulness as water tracers is based on their visibility at low concentrations, fluorescent properties, cost, relationship to pH and salinity, photochemical decay, and resistance to absorption on mineral and organic surfaces. A comparison of eight fluorescent dyes was conducted by [Smart et Laidlaw 1977]. Fluorescein (green) and Rhodamine WT (orange) are preferred and can be combined.

- **Dissolved organic matter (DOM)** can be distinguished by its fluorescence spectra, which help track biogeochemical processes, circulation patterns, and water masses. DOM fluorimetry is a useful tracer for freshwater such as polar waters and is even used to track their origin [Gonçalves-Araujo et al., 2016]. In the same study, the evolution of fluorescence properties was correlated with apparent oxygen utilization in deep waters, demonstrating that fluorescence can also be used to study mixing dynamics and DOM turnover processes [Gonçalves-Araujo et al., 2016].

- **Non-marine objects**, such as plastic debris (from micro to macro), oil spills, or human bodies can be more easily tracked using fluorescent cameras and sensors than visible cameras [Misra et al. 2025].

#### Environmental parameter measurement
**Chemical sensing through quenching:** Fluorescence intensity depends strongly on environmental parameters through quenching processes. These relationships are exploited in optodes for measuring:
- **Dissolved oxygen:** Higher oxygen concentrations quench fluorescence intensity (see more in the [oxygen section](#background_oxygene))
- **pH monitoring:** Fluorescent dyes sensitive to hydrogen ion concentration serve as proxies for pH measurements [Szapoczka et al. 2023]

The environmental sensitivity that initially appears as a limitation becomes an advantage when properly calibrated, enabling multi-parameter sensing with single fluorescent systems. Planar optodes are used to study O2 and pH evolution through sediment to quantify bioturbation [Larsen et al. 2011]. Luminophores, i.e., colored sand grains visible under UV light, are also typically used in bioturbation studies [Mahaut et Graf 1987].

### Units and standards

Calibration of fluorescence sensors depends on the parameters of interest and is challenging, especially when dealing with concentrations of living particles such as phytoplankton. 

- For chlorophyll-a fluorescence measurements, primary calibration was performed using extracted chlorophyll solutions of known concentration, prepared from cultured phytoplankton or spinach extracts dissolved in acetone [Holm-Hansen et al., 1965; Lorenzen, 1966]. Today, standards mostly include stable fluorescent dyes such as Rhodamine WT [see the characteristics of a typical [WiMo instrument](https://nke-instrumentation.fr/wp-content/uploads/2025/05/Brochure-WiMo-frw.pdf)] and Fluorescein. 

- CDOM fluorescence is conventionally calibrated against quinine sulfate dihydrate solutions, with results expressed in Quinine Sulfate Units (QSU) or parts per billion (ppb) quinine sulfate equivalents. Instrument-specific calibration coefficients must account for temperature dependencies, non-linear response at high concentrations, and potential quenching effects. Regular verification using certified reference materials and inter-comparison exercises between instruments are essential practices to maintain measurement accuracy and facilitate integration of datasets across different platforms and temporal scales.

## Sensing
To measure fluorophore concentration or environmental variation, three types of fluorimetry are used, each with distinct principles and applications in oceanographic instrumentation.

### Fluorescence intensity change 

Fluorescence intensity measurements are the most common approach for measuring chlorophyll-a and colored dissolved organic matter (CDOM) concentrations.

:::{dropdown} ▶ Principle

Fluorescence intensity measurements rely on detecting the transmitted light intensity when a sample is excited at a specific wavelength. The basic principle follows the Beer-Lamber's law :

$I_f = K \cdot I_0 \cdot (1- e^{- \epsilon c L})$

Where:
- $I_f$ = fluorescence intensity
- $K$ = quantum yield 
- $I_0$ = incident light intensity
- $\epsilon$ = molar absorption coefficient
- $c$ = fluorophore concentration
- $L$ = path length

For low concentrations, fluorescence intensity is proportional to concentration. 

```{figure} ../_static/fluo/fluo_geometry.png
:width: 500px
:name: fluo_geometry
:align: center

Figure 5 - Typical fluorometer geometry showing excitation light source, sample, and detector positioned at 90° angle
```

Typical excitation and emission wavelength are:
- **Chlorophyll-a**: Excitation ~470 nm (blue), Emission ~694 nm (red). Based on NKE instrument.
- **CDOM**: Excitation ~350-380 nm (UV), Emission ~450-520 nm (blue-green). From Aquared and NKE instrument.
:::

:::{dropdown} ▶ Different types of sensors

Fluorometers are most commonly LED-based for their long lifetime and low power consumption. The measurement geometry is typically at 90° to minimize interference from excitation light and scattered light.

**Photodetector options:**
- **Photodiodes**: Fast response, good sensitivity, most common choice
- **Photomultiplier tubes (PMTs)**: High sensitivity but require high voltage
- **Silicon detectors with filters**: Cost-effective for specific wavelengths

**Filter selection:**
- **Excitation filters**: Narrow bandpass to select specific excitation wavelengths
- **Emission filters**: Long-pass or bandpass to isolate fluorescence signal
- **Dichroic mirrors**: Separate excitation and emission paths efficiently

**Commercial examples:**
- Turner Designs C3 Submersible Fluorometer
- WET Labs ECO series
- Sea-Bird Scientific fluorometers
:::

:::{dropdown} ▶ Typical characteristics

For chlorophyll-a fluorometers:
- **Detection limit**: 0.01-0.03 μg/L
- **Dynamic range**: 0-400 μg/L (high-gain mode)
- **Resolution**: 0.01 μg/L
- **Response time**: <1 second
- **Accuracy**: ±5% or ±0.02 μg/L (whichever is greater)
- **Power consumption**: 0.1-2 W depending on design

For CDOM fluorometers:
- **Detection limit**: 0.1-1 ppb quinine sulfate units (QSU)
- **Dynamic range**: 0-300 QSU
- **Resolution**: 0.1 QSU
- **Excitation/Emission**: 370/460 nm typical
:::

:::{dropdown} ▶ Advantages and limitations

**Advantages:**
- Simple and robust measurement principle
- Fast response time suitable for real-time monitoring
- Well-established calibration standards
- Wide dynamic range for most oceanographic applications
- Low power consumption in LED-based designs

**Limitations:**
- Affected by temperature variations (typically -1% to -2% per °C)
- Subject to quenching effects (oxygen, pH, salinity)
- Fouling in long-term deployments affects accuracy
- Cannot distinguish between different types of particles directly
- Requires regular cleaning and calibration
:::

### Fluorescent spectroscopy

Spectroscopic measurements provide spectral information about fluorescent materials, enabling identification and differentiation of various fluorophores.

:::{dropdown} ▶ Principle

Fluorescence spectroscopy measures the complete emission spectrum of a sample when excited at one or more wavelengths, or the excitation spectrum when monitoring at a fixed emission wavelength. This provides a "fingerprint" of the fluorescent material.

**Excitation-Emission Matrices (EEMs):**
- Systematic measurement across multiple excitation and emission wavelengths
- Creates 3D plots showing fluorescence intensity as a function of both wavelengths
- Enables identification of multiple fluorophore components in complex samples

**Parallel Factor Analysis (PARAFAC):**
- Mathematical decomposition of EEM data
- Separates overlapping fluorescence signals
- Widely used for CDOM characterization in marine environments

The spectroscopic approach enables:
- Identification of phytoplankton species based on pigment composition
- Characterization of dissolved organic matter quality
- Detection of pollutants and contaminants
- Monitoring of protein-like vs humic-like fluorescence
:::

:::{dropdown} ▶ Different types of sensors

**Miniature spectrometers:**
- Ocean Optics series (USB2000, HR series)
- Hamamatsu mini-spectrometers
- Typical range: 200-1100 nm
- Resolution: 0.1-2 nm depending on model

**LED arrays for excitation:**
- Multiple wavelength LEDs for pseudo-spectroscopic measurements
- Typical wavelengths: 280, 310, 340, 380, 410, 445, 470, 505 nm
- Sequential excitation with single emission detector

**Filter wheels:**
- Motorized filter changers for multi-wavelength capability
- Slower than LED arrays but higher optical quality
- Better wavelength precision and isolation

**Fiber optic systems:**
- Remote sensing capability
- Bifurcated fibers separate excitation and emission paths
- Suitable for in-situ deployment
:::

:::{dropdown} ▶ Typical characteristics

**Mini-spectrometer systems:**
- **Wavelength range**: 200-1100 nm (typical)
- **Resolution**: 0.3-2 nm FWHM
- **Integration time**: 1 ms - 60 s
- **Signal-to-noise ratio**: >300:1
- **Size**: 89 x 64 x 34 mm (typical miniature unit)
- **Power**: 5V USB powered

**Multi-wavelength fluorometers:**
- **Number of channels**: 4-8 typical
- **Excitation wavelengths**: User-selectable LEDs
- **Detection**: Single PMT or photodiode
- **Measurement time**: 1-10 seconds per complete spectrum
:::

:::{dropdown} ▶ Advantages and limitations

**Advantages:**
- Species identification capability
- Multiple parameter measurement in single instrument
- Reduced interference from overlapping signals
- Better characterization of dissolved organic matter
- Can detect changes in phytoplankton community structure

**Limitations:**
- Higher complexity and cost
- Longer measurement times
- Requires sophisticated data processing
- More sensitive to optical alignment
- Higher power consumption
- Complex calibration procedures
:::

### Fluorescence Lifetime Measurements

Fluorescence lifetime measurements detect the decay time of fluorescence emission, providing additional discrimination capability and reduced interference from scattered light.

:::{dropdown} ▶ Principle

Fluorescence lifetime (τ) is the average time a molecule remains in the excited state before emitting a photon. It's measured by:

**Time-domain method:**
- Short pulse excitation (<< τ)
- Monitor exponential decay: $I(t) = I_0 e^{-t/τ}$
- Typical marine fluorophore lifetimes: 1-5 ns

**Frequency-domain method:**
- Sinusoidally modulated excitation light
- Measure phase shift and demodulation of emission
- More suitable for longer lifetimes (μs range)

Lifetime measurements offer several advantages:
- Independent of fluorophore concentration
- Less affected by optical path length variations
- Can distinguish fluorophores with similar spectra but different lifetimes
- Reduced interference from scattered excitation light
:::

:::{dropdown} ▶ Different types of sensors

**Time-Correlated Single Photon Counting (TCSPC):**
- Pulsed diode lasers or LEDs
- Single photon avalanche detectors (SPADs)
- Time-to-amplitude converters
- Nanosecond time resolution

**Gated detection systems:**
- Time-gated intensified cameras
- Programmable delay generators
- Suitable for lifetime imaging applications

**Phase fluorometry:**
- LED drivers with sinusoidal modulation
- Lock-in amplifiers for phase detection
- Simpler than TCSPC but lower time resolution

**Smartphone-based systems:**
- LED flash for pulsed excitation
- Camera sensor for decay detection
- Limited to longer lifetime fluorophores
:::

:::{dropdown} ▶ Typical characteristics

**TCSPC systems:**
- **Time resolution**: 10-50 ps
- **Lifetime range**: 100 ps - 100 μs
- **Count rate**: Up to 10⁷ counts/s
- **Accuracy**: ±1-5% depending on count statistics
- **Measurement time**: Seconds to minutes

**Phase fluorometry:**
- **Modulation frequency**: 1-200 MHz
- **Phase resolution**: 0.1-1°
- **Lifetime range**: 0.1-100 ns
- **Measurement time**: <1 second
:::

:::{dropdown} ▶ Advantages and limitations

**Advantages:**
- Concentration-independent measurements
- Excellent discrimination between fluorophores
- Minimal interference from scattering and background
- Can measure in highly turbid waters
- Potential for multi-exponential decay analysis

**Limitations:**
- Complex instrumentation and higher cost
- Requires fast electronics and detectors
- Limited to fluorophores with suitable lifetimes
- Data analysis complexity
- Lower signal levels require longer measurement times
- Not yet widely adopted in oceanographic applications
:::

````python
from IPython.display import HTML
import pandas as pd
import re

def create_interactive_fluorescence_table(df):
    """Create an interactive table with drag-and-drop columns and link icons for URLs"""
    
    # Function to convert URLs to clickable link icons
    def convert_url_to_link(text):
        if pd.isna(text):
            return "—"
        
        text_str = str(text)
        # Regex to find HTTP/HTTPS URLs
        url_pattern = r'(https?://[^\s<>"{}|\\^`\[\]]+)'
        
        def replace_url(match):
            url = match.group(1)
            return f'<a href="{url}" target="_blank" style="text-decoration: none; color: #3498db; font-size: 1.2em;">🔗</a>'
        
        return re.sub(url_pattern, replace_url, text_str)
    
    # Apply URL conversion to all columns
    df_converted = df.copy()
    for col in df_converted.columns:
        df_converted[col] = df_converted[col].apply(convert_url_to_link)
    
    # Generate column IDs for drag-and-drop
    column_ids = {col: f"col_{i}" for i, col in enumerate(df_converted.columns)}

    # Create table HTML with drag-and-drop functionality
    interactive_html = f"""
    <style>
    .table-container {{
        overflow-x: auto;
        margin: 20px 0;
    }}
    
    .interactive-fluorescence-table {{
        border-collapse: collapse;
        margin: 25px 0;
        font-size: 0.9em;
        font-family: sans-serif;
        min-width: 100%;
        box-shadow: 0 0 20px rgba(0, 0, 0, 0.15);
        background: white;
    }}
    
    .interactive-fluorescence-table thead th {{
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        color: white;
        font-weight: bold;
        padding: 15px 12px;
        text-align: left;
        position: relative;
        cursor: move;
        user-select: none;
        border: 1px solid #5a6c9d;
        transition: all 0.3s ease;
    }}
    
    .interactive-fluorescence-table thead th.sticky {{
        position: sticky;
        left: 0;
        z-index: 2;
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        box-shadow: 2px 0 5px -2px #ccc;
    }}
    
    .interactive-fluorescence-table thead th:hover {{
        background: linear-gradient(135deg, #5a6c9d 0%, #6a5a92 100%);
        transform: translateY(-2px);
        box-shadow: 0 4px 8px rgba(0,0,0,0.2);
    }}
    
    .interactive-fluorescence-table thead th.dragging {{
        opacity: 0.5;
        transform: rotate(5deg);
    }}
    
    .interactive-fluorescence-table thead th.drag-over {{
        background: linear-gradient(135deg, #f39c12 0%, #e67e22 100%);
        transform: scale(1.05);
    }}
    
    .interactive-fluorescence-table td {{
        padding: 12px 15px;
        border: 1px solid #dddddd;
        vertical-align: top;
        background: white;
    }}
    
    .interactive-fluorescence-table td.sticky {{
        position: sticky;
        left: 0;
        z-index: 2;
        background: white;
        box-shadow: 2px 0 5px -2px #ccc;
    }}
    
    .interactive-fluorescence-table tbody tr:nth-of-type(even) {{
        background-color: #f8f9fa;
    }}
    
    .interactive-fluorescence-table tbody tr:nth-of-type(even) td.sticky {{
        background-color: #f8f9fa;
    }}
    
    .interactive-fluorescence-table tbody tr:hover {{
        background-color: #e3f2fd;
        cursor: pointer;
    }}
    
    .interactive-fluorescence-table tbody tr:hover td.sticky {{
        background-color: #e3f2fd;
    }}
    
    .technology-cell {{
        font-weight: bold;
        color: #2c3e50;
        white-space: nowrap;
    }}
    
    .drag-hint {{
        position: absolute;
        top: -8px;
        right: 5px;
        font-size: 0.8em;
        opacity: 0.7;
    }}
    
    /* Cross link styling */
    .interactive-fluorescence-table a {{
        text-decoration: none;
        color: #e74c3c;
        font-weight: bold;
        font-size: 1.2em;
        transition: all 0.2s ease;
    }}
    
    .interactive-fluorescence-table a:hover {{
        color: #c0392b;
        transform: scale(1.3);
        text-shadow: 0 0 5px rgba(231, 76, 60, 0.5);
    }}
    </style>
    
    <div style="margin: 20px 0; padding: 15px; background: #f8f9fa; border-left: 4px solid #3498db; border-radius: 5px;">
        <strong>📋 Interactive Table Features:</strong><br>
        • <strong>Drag & Drop:</strong> Click and drag column headers to reorder columns<br>
        • <strong>Links:</strong> URLs are converted to clickable blue link (🔗) icons<br>
    </div>
    
    <div class="table-container">
    <table class="interactive-fluorescence-table" id="interactiveFluorescenceTable">
        <thead>
            <tr id="headerRow">
    """
    
    # Add index column header (make sticky) - changed to "Technology"
    interactive_html += f'<th draggable="true" data-column="index" id="header_index" class="sticky">Technology <span class="drag-hint">⇄</span></th>'
    
    # Add data column headers
    for col in df_converted.columns:
        col_id = column_ids[col]
        interactive_html += f'<th draggable="true" data-column="{col_id}" id="header_{col_id}">{col} <span class="drag-hint">⇄</span></th>'
    
    interactive_html += """
            </tr>
        </thead>
        <tbody id="tableBody">
    """
    
    # Add data rows
    for idx, (sensor_name, row) in enumerate(df_converted.iterrows()):
        interactive_html += '<tr>'
        
        # Add sensor name cell (make sticky) - using technology-cell class
        interactive_html += f'<td class="technology-cell sticky">{sensor_name}</td>'
        
        # Add data cells
        for col in df_converted.columns:
            value = row[col] if pd.notna(row[col]) else "—"
            interactive_html += f'<td>{value}</td>'
        
        interactive_html += '</tr>'
    
    interactive_html += """
        </tbody>
    </table>
    </div>
    
    <script>
    (function() {
        // Wait for DOM to be ready
        setTimeout(function() {
            console.log('Initializing interactive fluorescence table...');
            
            let draggedElement = null;
            let draggedColumnId = null;
            
            const headers = document.querySelectorAll('#interactiveFluorescenceTable thead th[draggable="true"]');
            console.log('Found', headers.length, 'draggable headers');
            
            headers.forEach(function(header) {
                header.addEventListener('dragstart', function(e) {
                    console.log('Drag start:', this.dataset.column);
                    draggedElement = this;
                    draggedColumnId = this.dataset.column;
                    this.classList.add('dragging');
                    e.dataTransfer.effectAllowed = 'move';
                    e.dataTransfer.setData('text/html', this.outerHTML);
                });
                
                header.addEventListener('dragend', function(e) {
                    console.log('Drag end');
                    this.classList.remove('dragging');
                    headers.forEach(function(h) {
                        h.classList.remove('drag-over');
                    });
                });
                
                header.addEventListener('dragover', function(e) {
                    if (e.preventDefault) {
                        e.preventDefault();
                    }
                    e.dataTransfer.dropEffect = 'move';
                    return false;
                });
                
                header.addEventListener('dragenter', function(e) {
                    if (this !== draggedElement) {
                        this.classList.add('drag-over');
                    }
                });
                
                header.addEventListener('dragleave', function(e) {
                    this.classList.remove('drag-over');
                });
                
                header.addEventListener('drop', function(e) {
                    if (e.stopPropagation) {
                        e.stopPropagation();
                    }
                    
                    console.log('Drop on:', this.dataset.column);
                    
                    if (draggedElement !== this) {
                        reorderColumns(draggedColumnId, this.dataset.column);
                    }
                    this.classList.remove('drag-over');
                    return false;
                });
            });
            
            function reorderColumns(draggedColId, targetColId) {
                console.log('Reordering columns:', draggedColId, 'to', targetColId);
                
                const table = document.getElementById('interactiveFluorescenceTable');
                const headerRow = document.getElementById('headerRow');
                const tbody = document.getElementById('tableBody');
                
                const draggedHeader = document.getElementById('header_' + draggedColId);
                const targetHeader = document.getElementById('header_' + targetColId);
                
                if (!draggedHeader || !targetHeader) {
                    console.error('Headers not found');
                    return;
                }
                
                const draggedIndex = Array.from(headerRow.children).indexOf(draggedHeader);
                const targetIndex = Array.from(headerRow.children).indexOf(targetHeader);
                
                console.log('Moving from index', draggedIndex, 'to', targetIndex);
                
                // Move header
                if (draggedIndex < targetIndex) {
                    headerRow.insertBefore(draggedHeader, targetHeader.nextSibling);
                } else {
                    headerRow.insertBefore(draggedHeader, targetHeader);
                }
                
                // Move corresponding data cells in each row
                const rows = tbody.querySelectorAll('tr');
                rows.forEach(function(row) {
                    const cells = row.children;
                    if (cells.length > draggedIndex && cells.length > targetIndex) {
                        const draggedCell = cells[draggedIndex];
                        const targetCell = cells[targetIndex];
                        
                        if (draggedIndex < targetIndex) {
                            row.insertBefore(draggedCell, targetCell.nextSibling);
                        } else {
                            row.insertBefore(draggedCell, targetCell);
                        }
                    }
                });
                
                console.log('Column reordering completed');
            }
            
        }, 100);
    })();
    </script>
    """
    
    return interactive_html

# Read and process the fluorescence sensor data
s = (
    pd.read_csv("sensors.csv", skiprows=0)
    .rename(columns=lambda x: x.strip())
    .set_index("sensor name")
)

# Create and display the interactive table
interactive_table_html = create_interactive_fluorescence_table(s)
HTML(interactive_table_html)
````
