# Zotero Library

This page presents the collaborative library of the DIY Oceanography BZH group.

## Usage

This library is automatically updated from the Zotero group. You can:

* **Search** in all fields: title, authors, abstract, keywords
* **Filter** by collection, document type or publication year  
* **Sort** results by title, date or author
* **Access** each reference directly in Zotero by clicking on the link

### Available Filters:
- **Collection**: Browse references organized by research topic or theme
- **Document type**: Filter by journal articles, books, reports, etc.
- **Year**: Narrow down by publication year
- **Search**: Real-time text search across all fields


## Contribute
