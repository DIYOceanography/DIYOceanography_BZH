### How to Use These Forms

1. **Choose your preferred language version** (French or English)
2. **Download the .docx file** using one of the download options:
   - **Primary button**: Downloads from the documentation site (if files are in `_static` folder)
   - **GitHub link**: Direct download from the GitHub repository (always works)
3. **Open the document** in Microsoft Word or compatible software
4. **Fill out the form** with your project-specific requirements
5. **Save and share** with your team or stakeholders

These forms are designed to help you systematically define and document the metrological requirements for your DIY oceanographic instrumentation projects, ensuring that all measurement needs are properly planned and specified.

### Alternative Access

If the download buttons don't work, you can also:
- **Browse the repository**: Visit the [metrology folder on GitHub](https://github.com/DIYOceanography/DIYOceanography_BZH/tree/main/docs/metrology) to view and download the files
- **Right-click and save**: Right-click on the GitHub direct links above and select "Save link as..."
