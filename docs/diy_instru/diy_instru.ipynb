# DIY instruments

## What is a Complete DIY Project?

A complete DIY (Do-It-Yourself) project in oceanographic instrumentation typically includes:

1. **Hardware Components**
    - Sensor elements and electronics
    - Waterproof housing and mechanical parts
    - Assembly instructions and diagrams

2. **Software Elements** 
    - Data acquisition code
    - Calibration procedures
    - Data processing scripts

3. **Documentation**
    - Build instructions
    - Parts list with costs
    - Testing procedures
    - Validation results

4. **Validation**
    - Comparison with other instruments
    - Field testing results
    - Known limitations

A project is considered complete when all these elements are publicly available and have been validated in real conditions. 
It should also be clearly stated the limitations and advantages of the instruments. The reparability and environmental impact is secondary but nevertheless important. 
An example on how assess the carbon footprint of a reusable glassware is automated in EcoLabware [Sophie et al 2025].

### References

- Sophie Schbath, Max-Henri Chanut, Marianne De Paepe (2025). EcoLabware: Assessing the carbon footprint of single-use plastic ware versus reusable glassware.

## CTD

This table lists DIY CTD (Conductivity, Temperature, Depth) instruments with their specifications and capabilities.

```python
from IPython.display import HTML
import pandas as pd
import re

# Read the CTD CSV file
df = pd.read_csv("DIY-instruments-all/CTD-Tableau 1.csv", 
                 sep=';',
                 na_values=['', 'NaN', 'nan']).set_index("Project")

# Clean column names
df = df.rename(columns=lambda x: x.strip())

def convert_url_to_cross(text):
    """Convert URLs to clickable cross symbols"""
    if pd.isna(text):
        return "—"
    
    text_str = str(text)
    
    # URL pattern
    url_pattern = r'https?://[^\s<>"{}|\\^`\[\]]+'
    
    def replace_url(match):
        url = match.group(0)
        return f'<a href="{url}" target="_blank" title="{url}">🔗</a>'
    
    # Replace URLs with cross symbols
    converted = re.sub(url_pattern, replace_url, text_str)
    
    return converted

# Create a copy of the dataframe for URL conversion
df_converted = df.copy()

# Apply URL conversion to all columns
for col in df_converted.columns:
    df_converted[col] = df_converted[col].apply(convert_url_to_cross)

# Generate column IDs for drag-and-drop
column_ids = {col: f"col_{i}" for i, col in enumerate(df_converted.columns)}

# Create table HTML with drag-and-drop functionality
interactive_html = f"""
    <style>
    .table-container {{
        overflow-x: auto;
        max-width: 100%;
    }}
    
    .interactive-ctd-table {{
        border-collapse: collapse;
        margin: 25px 0;
        font-size: 0.9em;
        font-family: sans-serif;
        min-width: 100%;
        box-shadow: 0 0 20px rgba(0, 0, 0, 0.15);
        background: white;
    }}
    
    .interactive-ctd-table thead th {{
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        color: #ffffff;
        font-weight: bold;
        padding: 12px 15px;
        text-align: left;
        border: 1px solid #dddddd;
        cursor: move;
        position: relative;
        user-select: none;
    }}
    
    .interactive-ctd-table thead th.sticky {{
        position: sticky;
        left: 0;
        z-index: 3;
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        box-shadow: 2px 0 5px -2px rgba(0,0,0,0.3);
    }}
    
    .interactive-ctd-table thead th:hover {{
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        transform: scale(1.05);
    }}
    
    .interactive-ctd-table td {{
        padding: 12px 15px;
        border: 1px solid #dddddd;
        vertical-align: top;
        background: white;
    }}
    
    .interactive-ctd-table td.sticky {{
        position: sticky;
        left: 0;
        z-index: 2;
        background: #f1f8ff;
        box-shadow: 2px 0 5px -2px rgba(0,0,0,0.3);
    }}
    
    .interactive-ctd-table tbody tr:nth-of-type(even) {{
        background-color: #f8f9fa;
    }}
    
    .interactive-ctd-table tbody tr:nth-of-type(even) td.sticky {{
        background-color: #e8f4ff;
    }}
    
    .interactive-ctd-table tbody tr:hover {{
        background-color: #e3f2fd;
        cursor: pointer;
    }}
    
    .interactive-ctd-table tbody tr:hover td.sticky {{
        background-color: #d1e7fd;
    }}
    
    .technology-cell {{
        font-weight: bold;
        background-color: #f1f8ff !important;
    }}
    
    .drag-hint {{
        font-size: 0.8em;
        opacity: 0.7;
        margin-left: 5px;
    }}
    
    .dragging {{
        opacity: 0.5;
        transform: scale(0.95);
    }}
    
    .drag-over {{
        background: linear-gradient(135deg, #6FA8DC 0%, #8FBC8F 100%) !important;
        transform: scale(1.02);
    }}
    
    .interactive-ctd-table a {{
        color: #e74c3c;
        text-decoration: none;
        font-size: 1.2em;
        display: inline-block;
        transition: all 0.3s ease;
    }}
    
    .interactive-ctd-table a:hover {{
        color: #c0392b;
        transform: scale(1.3);
        text-shadow: 0 0 5px rgba(231, 76, 60, 0.5);
    }}
    </style>
    
    <div class="table-container">
        <table class="interactive-ctd-table" id="interactiveCTDTable">
            <thead>
                <tr id="headerRow">
"""

# Add index column header (make sticky)
interactive_html += f'<th draggable="true" data-column="index" id="header_index" class="sticky">Project <span class="drag-hint">⇄</span></th>'

# Add data column headers
for col in df_converted.columns:
    col_id = column_ids[col]
    interactive_html += f'<th draggable="true" data-column="{col_id}" id="header_{col_id}">{col} <span class="drag-hint">⇄</span></th>'

interactive_html += """
                </tr>
            </thead>
            <tbody id="tableBody">
"""

# Add data rows
for idx, (project, row) in enumerate(df_converted.iterrows()):
    interactive_html += '<tr>'
        
    # Add project cell with sticky class
    interactive_html += f'<td class="technology-cell sticky">{project}</td>'
        
    # Add data cells
    for col in df_converted.columns:
        value = row[col] if pd.notna(row[col]) else "—"
        interactive_html += f'<td>{value}</td>'
        
    interactive_html += '</tr>'
    
interactive_html += """
            </tbody>
        </table>
    </div>
    
    <script>
    (function() {
        // Wait for DOM to be ready
        setTimeout(function() {
            console.log('Initializing CTD interactive table...');

            let draggedElement = null;
            let draggedColumnId = null;

            const headers = document.querySelectorAll('#interactiveCTDTable thead th[draggable="true"]');

            headers.forEach(function(header) {
                header.addEventListener('dragstart', function(e) {
                    console.log('Drag start:', this.dataset.column);
                    draggedElement = this;
                    draggedColumnId = this.dataset.column;
                    this.classList.add('dragging');

                    e.dataTransfer.effectAllowed = 'move';
                    e.dataTransfer.setData('text/html', this.outerHTML);
                });

                header.addEventListener('dragend', function(e) {
                    console.log('Drag end');
                    this.classList.remove('dragging');
                    
                    // Remove drag-over class from all headers
                    headers.forEach(function(h) {
                        h.classList.remove('drag-over');
                    });
                });

                header.addEventListener('dragover', function(e) {
                    if (draggedElement && draggedElement !== this) {
                        e.preventDefault();
                        e.dataTransfer.dropEffect = 'move';
                        this.classList.add('drag-over');
                    }
                });

                header.addEventListener('dragleave', function(e) {
                    this.classList.remove('drag-over');
                });

                header.addEventListener('drop', function(e) {
                    e.preventDefault();
                    this.classList.remove('drag-over');
                    
                    if (draggedElement && draggedElement !== this) {
                        console.log('Dropping', draggedColumnId, 'onto', this.dataset.column);
                        
                        const draggedIndex = Array.from(headers).indexOf(draggedElement);
                        const targetIndex = Array.from(headers).indexOf(this);
                        
                        reorderColumns(draggedIndex, targetIndex);
                    }
                });
            });

            function reorderColumns(draggedIndex, targetIndex) {
                console.log('Reordering columns:', draggedIndex, 'to', targetIndex);
                
                const table = document.getElementById('interactiveCTDTable');
                const headerRow = table.querySelector('#headerRow');
                const bodyRows = table.querySelectorAll('#tableBody tr');

                // Move header
                const headerCells = Array.from(headerRow.children);
                const draggedHeader = headerCells[draggedIndex];
                
                if (targetIndex < draggedIndex) {
                    headerRow.insertBefore(draggedHeader, headerCells[targetIndex]);
                } else {
                    headerRow.insertBefore(draggedHeader, headerCells[targetIndex + 1]);
                }

                // Move body cells
                bodyRows.forEach(function(row) {
                    const cells = Array.from(row.children);
                    const draggedCell = cells[draggedIndex];
                    
                    if (targetIndex < draggedIndex) {
                        row.insertBefore(draggedCell, cells[targetIndex]);
                    } else {
                        row.insertBefore(draggedCell, cells[targetIndex + 1]);
                    }
                });
            }
        }, 100);
    })();
    </script>
"""
# Display the interactive table
HTML(interactive_html)
```

```python
# Read the CSV file
df = pd.read_csv("DIY-instruments-all/Buoy-Tableau 1.csv", 
                 sep=';',
                 na_values=['', 'NaN', 'nan']).set_index("Project")

# Clean column names
df = df.rename(columns=lambda x: x.strip())

def convert_url_to_cross(text):
    """Convert URLs to clickable cross symbols"""
    if pd.isna(text):
        return "—"
    
    text_str = str(text)
    
    # URL pattern
    url_pattern = r'https?://[^\s<>"{}|\\^`\[\]]+'
    
    def replace_url(match):
        url = match.group(0)
        return f'<a href="{url}" target="_blank" title="{url}">🔗</a>'
    
    # Replace URLs with cross symbols
    converted = re.sub(url_pattern, replace_url, text_str)
    
    return converted

# Create a copy of the dataframe for URL conversion
df_converted = df.copy()

# Apply URL conversion to all columns
for col in df_converted.columns:
    df_converted[col] = df_converted[col].apply(convert_url_to_cross)

# Generate column IDs for drag-and-drop
column_ids = {col: f"col_{i}" for i, col in enumerate(df_converted.columns)}

# Create table HTML with drag-and-drop functionality
interactive_html = f"""
    <style>
    .table-container {{
        overflow-x: auto;
        max-width: 100%;
    }}
    
    .interactive-buoy-table {{
        border-collapse: collapse;
        margin: 25px 0;
        font-size: 0.9em;
        font-family: sans-serif;
        min-width: 100%;
        box-shadow: 0 0 20px rgba(0, 0, 0, 0.15);
        background: white;
    }}
    
    .interactive-buoy-table thead th {{
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        color: #ffffff;
        font-weight: bold;
        padding: 12px 15px;
        text-align: left;
        border: 1px solid #dddddd;
        cursor: move;
        position: relative;
        user-select: none;
    }}
    
    .interactive-buoy-table thead th.sticky {{
        position: sticky;
        left: 0;
        z-index: 3;
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        box-shadow: 2px 0 5px -2px rgba(0,0,0,0.3);
    }}
    
    .interactive-buoy-table thead th:hover {{
        background: linear-gradient(135deg, #4A7A9E 0%, #5A92BA 100%);
        transform: scale(1.05);
    }}
    
    .interactive-buoy-table td {{
        padding: 12px 15px;
        border: 1px solid #dddddd;
        vertical-align: top;
        background: white;
    }}
    
    .interactive-buoy-table td.sticky {{
        position: sticky;
        left: 0;
        z-index: 2;
        background: #f1f8ff;
        box-shadow: 2px 0 5px -2px rgba(0,0,0,0.3);
    }}
    
    .interactive-buoy-table tbody tr:nth-of-type(even) {{
        background-color: #f8f9fa;
    }}
    
    .interactive-buoy-table tbody tr:nth-of-type(even) td.sticky {{
        background-color: #e8f4ff;
    }}
    
    .interactive-buoy-table tbody tr:hover {{
        background-color: #e3f2fd;
        cursor: pointer;
    }}
    
    .interactive-buoy-table tbody tr:hover td.sticky {{
        background-color: #d1e7fd;
    }}
    
    .technology-cell {{
        font-weight: bold;
        background-color: #f1f8ff !important;
    }}
    
    .drag-hint {{
        font-size: 0.8em;
        opacity: 0.7;
        margin-left: 5px;
    }}
    
    .dragging {{
        opacity: 0.5;
        transform: scale(0.95);
    }}
    
    .drag-over {{
        background: linear-gradient(135deg, #6FA8DC 0%, #8FBC8F 100%) !important;
        transform: scale(1.02);
    }}
    
    .interactive-buoy-table a {{
        color: #e74c3c;
        text-decoration: none;
        font-size: 1.2em;
        display: inline-block;
        transition: all 0.3s ease;
    }}
    
    .interactive-buoy-table a:hover {{
        color: #c0392b;
        transform: scale(1.3);
        text-shadow: 0 0 5px rgba(231, 76, 60, 0.5);
    }}
    </style>

    <div class="table-container">
        <table class="interactive-buoy-table" id="interactiveBuoyTable">
            <thead>
                <tr id="headerRow">
"""

# Add index column header (make sticky)
interactive_html += f'<th draggable="true" data-column="index" id="header_index" class="sticky">Project <span class="drag-hint">⇄</span></th>'

# Add data column headers
for col in df_converted.columns:
    col_id = column_ids[col]
    interactive_html += f'<th draggable="true" data-column="{col_id}" id="header_{col_id}">{col} <span class="drag-hint">⇄</span></th>'

interactive_html += """
                </tr>
            </thead>
            <tbody id="tableBody">
"""

# Add data rows
for idx, (project, row) in enumerate(df_converted.iterrows()):
    interactive_html += '<tr>'
        
    # Add project cell with sticky class
    interactive_html += f'<td class="technology-cell sticky">{project}</td>'
        
    # Add data cells
    for col in df_converted.columns:
        value = row[col] if pd.notna(row[col]) else "—"
        interactive_html += f'<td>{value}</td>'
        
    interactive_html += '</tr>'
    
interactive_html += """
            </tbody>
        </table>
    </div>
    
    <script>
    (function() {
        // Wait for DOM to be ready
        setTimeout(function() {
            console.log('Initializing Buoy interactive table...');

            let draggedElement = null;
            let draggedColumnId = null;

            const headers = document.querySelectorAll('#interactiveBuoyTable thead th[draggable="true"]');

            headers.forEach(function(header) {
                header.addEventListener('dragstart', function(e) {
                    console.log('Drag start:', this.dataset.column);
                    draggedElement = this;
                    draggedColumnId = this.dataset.column;
                    this.classList.add('dragging');

                    e.dataTransfer.effectAllowed = 'move';
                    e.dataTransfer.setData('text/html', this.outerHTML);
                });

                header.addEventListener('dragend', function(e) {
                    console.log('Drag end');
                    this.classList.remove('dragging');
                    
                    // Remove drag-over class from all headers
                    headers.forEach(function(h) {
                        h.classList.remove('drag-over');
                    });
                });

                header.addEventListener('dragover', function(e) {
                    if (draggedElement && draggedElement !== this) {
                        e.preventDefault();
                        e.dataTransfer.dropEffect = 'move';
                        this.classList.add('drag-over');
                    }
                });

                header.addEventListener('dragleave', function(e) {
                    this.classList.remove('drag-over');
                });

                header.addEventListener('drop', function(e) {
                    e.preventDefault();
                    this.classList.remove('drag-over');
                    
                    if (draggedElement && draggedElement !== this) {
                        console.log('Dropping', draggedColumnId, 'onto', this.dataset.column);
                        
                        const draggedIndex = Array.from(headers).indexOf(draggedElement);
                        const targetIndex = Array.from(headers).indexOf(this);
                        
                        reorderColumns(draggedIndex, targetIndex);
                    }
                });
            });

            function reorderColumns(draggedIndex, targetIndex) {
                console.log('Reordering columns:', draggedIndex, 'to', targetIndex);
                
                const table = document.getElementById('interactiveBuoyTable');
                const headerRow = table.querySelector('#headerRow');
                const bodyRows = table.querySelectorAll('#tableBody tr');

                // Move header
                const headerCells = Array.from(headerRow.children);
                const draggedHeader = headerCells[draggedIndex];
                
                if (targetIndex < draggedIndex) {
                    headerRow.insertBefore(draggedHeader, headerCells[targetIndex]);
                } else {
                    headerRow.insertBefore(draggedHeader, headerCells[targetIndex + 1]);
                }

                // Move body cells
                bodyRows.forEach(function(row) {
                    const cells = Array.from(row.children);
                    const draggedCell = cells[draggedIndex];
                    
                    if (targetIndex < draggedIndex) {
                        row.insertBefore(draggedCell, cells[targetIndex]);
                    } else {
                        row.insertBefore(draggedCell, cells[targetIndex + 1]);
                    }
                });
            }
        }, 100);
    })();
    </script>
"""
# Display the interactive table
HTML(interactive_html)
```
